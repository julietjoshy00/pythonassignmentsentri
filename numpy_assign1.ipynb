```python
import numpy as np

# Step 1: 1D array for week 1
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# Step 2: shape, data type, number of elements
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
# Step 3: Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temperatures_f)

# max, min, mean
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean:", temperatures_w1.mean())
```

```python
# Step 4: slicing
print("First three days:", temperatures_w1[:3])
print("Weekend (last two days):", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

```python
# Step 5: 2D array, each row is one week
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
# Step 6: inspect
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total elements:", temperatures.size)

# each week
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])

# weekends (last two days) of both weeks
print("Weekend of both weeks:")
print(temperatures[:, -2:])
```

```python
import pandas as pd

# Step 1: create Series
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
# Step 2: indexing and slicing

# integer position of the 1st rank student
print("1st rank mark:", marks.iloc[0])

# loc with labels for top 3 ranks
print("Top 3 ranks:")
print(marks.loc['Rank1':'Rank3'])

# iloc for 3rd rank student
print("3rd rank mark:", marks.iloc[2])

# boolean mask, marks greater than 90
print("Ranks with marks above 90:")
print(marks[marks > 90])
```

```python
# Step 3: manipulating

# change 1st rank mark to 100
marks['Rank1'] = 100
print(marks)

# remove the last rank
marks = marks.drop('Rank5')
print(marks)

# CGPA = marks / 10
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
# Step 1: create DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python
# Step 2: basic information
print("Head:")
print(transactions.head())

print("\nTail:")
print(transactions.tail())

print("\nShape:", transactions.shape)
print("Column names:", list(transactions.columns))
print("\nData types:")
print(transactions.dtypes)

print("\nInfo:")
transactions.info()
```

```python
# only two columns
print(transactions[['ProductCategory', 'Amount']])

# last 3 columns using iloc
print(transactions.iloc[:, -3:])

# Region is North and Amount greater than 200
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

# value counts
print(transactions['ProductCategory'].value_counts())

# unique regions
print(transactions['Region'].unique())

# mean amount for each region
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Step 3: manipulating

# change Amount of transaction 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)

# add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)

# remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)

# delete the Discount column
transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
