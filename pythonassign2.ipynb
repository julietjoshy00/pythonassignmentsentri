```python
# 1. List creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Arun", "Bala", "Charu", "Divya", "Esha"]
print("age_list :", age_list)
print("name_list:", name_list)
```

```python
# 2. List operations / modifications
name_list.append("Yazhini")        # a. add at the end
print("\n2a. After append:", name_list)

age_list.insert(2, 30)             # b. insert 30 at index 2
print("2b. After insert:", age_list)

name_list.remove("Yazhini")        # c. remove by value
print("2c. After remove:", name_list)

popped = age_list.pop()            # d. removes and returns the last element
print("2d. Popped element:", popped, "| age_list now:", age_list)

age_list.extend([29, 30, 26])      # e. add multiple elements at the end
print("2e. After extend:", age_list)

age_list.sort(reverse=True)        # f. descending order
print("2f. Sorted descending:", age_list)

# g. max, min, sum
print("2g. Max age:", max(age_list))
print("    Min age:", min(age_list))
print("    Sum of ages:", sum(age_list))
```

```python
# 3. Accessing list elements
print("\n3a. First element:", name_list[0])
print("3b. Last element:", name_list[-1])
print("3c. Index 2 to 4:", name_list[2:5])
print("3d. Reverse order:", name_list[::-1])
```

```python
# a. Create dictionary
student_marks = {
    "Arun": 78,
    "Bala": 65,
    "Charu": 92,
    "Divya": 55,
    "Esha": 88,
}
```

```python
# b. Access a specific student's mark
print("b. Charu's mark:", student_marks["Charu"])

# c. Add a new student
student_marks["Janani"] = 80
print("c. After adding Janani:", student_marks)

# d. Update an existing student's mark
student_marks["Bala"] = 82
print("d. After updating Bala:", student_marks)

# e. keys(), values(), items()
print("e. Keys  :", list(student_marks.keys()))
print("   Values:", list(student_marks.values()))
print("   Items :", list(student_marks.items()))
```

```python
# a. Create a set from a list with duplicates
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print("a. my_set:", my_set)
```

Explanation: A set stores only UNIQUE elements, so the repeated 'a' and 'i' are removed automatically. The result has 5 elements: a, e, i, o, u.Sets are also UNORDERED, so the elements may print in a different order each time you run the program.

```python
# b. Try to change an element using an index
try:
    my_set[4] = 's'
except TypeError as error:
    print("b. Error:", error)
```

Explanation: A set is unordered and has no index positions, so my_set[4] is not allowed. Sets do not support indexing or item assignment, which raises: TypeError: 'set' object does not support item assignment.To "change" a value, remove the old one and add the new one instead:

```python
my_set.remove('u')
my_set.add('s')
print("   After remove('u') and add('s'):", my_set)
```

```python
# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d. Union and intersection
print("d. Union       :", set1 | set2)
print("   Intersection:", set1 & set2)
```
